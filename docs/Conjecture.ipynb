\paragraph{Decision and next steps.}
The NeurIPS ``Generative AI in Finance'' workshop decision cited two gaps: (i) insufficient theoretical justification and (ii) not enough empirical evidence. Item (ii) can be addressed with additional experiments and robustness checks. For (i), 


The target statement is:

$\textbf{Conjecture:}$ 
$$
\kappa\!\big(\operatorname{Cov}(X)\big)\ \ge\ \kappa\!\big(\operatorname{Cov}(\tilde{X})\big),
\quad
\kappa(\Sigma)=\frac{\lambda_{\max}(\Sigma)}{\lambda_{\min}(\Sigma)},
$$
where $\tilde{X}$ is sampled from a DDPM fit on $X$.


A general proof is not in hand (and may fail without additional conditions). A plausible route is to impose linear--Gaussian structure or approximate the score with a linear (OLS) model, yielding tractable covariance updates so one can track eigenvalue shrinkage/anisotropy along the denoising chain. I will pursue this theoretical direction in parallel with expanded empirical validation.
